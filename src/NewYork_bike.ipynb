```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## **Import, combine and create the final the DataFrames**

```python
df1 = pd.read_csv('Downloads/bq-results-20260930-094439-1790761550657.csv')
df1.head(3)
```

```python
df2 = pd.read_csv('Downloads/bq-results-20260930-095624-1790762205631.csv')
df2.head(3)
```

```python
df3 = pd.read_csv('Downloads/bq-results-20260930-120628-1790770004796.csv')
df3.head(3)
```

```python
df4 = pd.read_csv('Downloads/bq-results-20260930-120917-1790770296702.csv')
df4.head(3)
```

```python
df5 = pd.read_csv('Downloads/bq-results-20260930-121351-1790770491288.csv')
df5.head(3)
```

```python
df6 = pd.read_csv('Downloads/bq-results-20260930-122217-1790770964762.csv')
df6.head(3)
```

```python
merged_df = pd.concat([df1,df2,df3,df4,df5],ignore_index=True)
merged_df.shape
```

```python
merged_df.head()
```

```python
df6.head()
```

```python
df6.columns
```

```python
# select the desired features to be combined with the merged df.
selected_columns = ['name','latitude','longitude','region_id','rental_methods']
df6_start = df6[selected_columns]
df6_start.head()
```

```python
# Merge the dataframes based on the start station name and then change the columns names of lat and long to indicate start.
ride_start_df = merged_df.merge(df6_start, how='inner',left_on='start_station_name', right_on='name')
ride_start_df.head()
```

```python
ride_start_df = ride_start_df.rename(columns={'latitude':'latitude_start',
                                              'longitude': 'longitude_start'})
```

```python
ride_start_df = ride_start_df.drop(columns='name')
ride_start_df.shape
```

```python
# extract the coordinates of the stop stations.
df6_stop = df6_start.drop(columns=['region_id','rental_methods'])
df6_stop.shape
```

```python
# Add the positional coordinates of the ride stop stations.
start_stop_df = ride_start_df.merge(df6_stop, how='inner', left_on='end_station_name', right_on='name')
```

```python
start_stop_df = start_stop_df.drop(columns='name').rename(columns={'latitude':'latitude_stop',
                                                                   'longitude':'longitude_stop'})
```

```python
df_final = start_stop_df.copy()
df_final.columns
```

```python
import numpy as np

R = 6371.0

lat1 = np.radians(df_final['latitude_start'])
lon1 = np.radians(df_final['longitude_start'])
lat2 = np.radians(df_final['latitude_stop'])
lon2 = np.radians(df_final['longitude_stop'])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2)**2
    + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
)

c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

df_final['distance_km'] = R * c
```

```python
df_final.head(2)
```

```python
# drop the latitude and longtide columns.
drop_columns = ['latitude_start','longitude_start','latitude_stop','longitude_stop']
df_clean = df_final.drop(columns=drop_columns)
df_clean.head(3)
```

## **Data Wrangling**

```python
# Observe the datatype of each columns.
df_clean.info()
```

```python
# convert the starttime and stoptimes to datetime formats.
df_clean['starttime'] = pd.to_datetime(df_clean['starttime'])
df_clean['stoptime'] = pd.to_datetime(df_clean['stoptime'])
```

```python
# extract the hour of the bike started.
df_clean['hour'] = df_clean['starttime'].dt.hour
```

```python
df_clean.head()
```

```python
# Chech if there are nulls and duplicates.
df_clean.isna().sum()
```

```python
df_clean.duplicated().sum()
```

```python
# Drop the duplicates
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.duplicated().sum()
```

## **Analysis**

### **Which station had the highest demand of bikes in 2014?**

```python
# count the number of trips per start_station_name and order descendingly.
uses_per_station = df_clean.groupby('start_station_name')['starttime'].count().reset_index(name='num_trips').sort_values(by='num_trips', ascending=False)
uses_per_station.head(10)
```

```python
uses_per_station_end = df_clean.groupby('end_station_name')['stoptime'].count().reset_index(name='num_trips')
uses_per_station_end.sort_values(by='num_trips', ascending=False).head(10)
```

```python
# show the distribution of the number of bikes demanded
sns.histplot(uses_per_station, x='num_trips', bins= 20, kde=True)
```

```python
# plot the top10 stations with the highest number of bike demands.
plt.figure(figsize=(5,4))
sns.barplot(uses_per_station.head(10), x='start_station_name', y='num_trips', color='orange')
plt.xticks(rotation=90)
plt.show()
```

### Stations having more than 35000 bike usage per year can be regarded as the most busy station. So, there is a high demand here. 

## **To which station do users ride mostly from 8 Ave & W 31 St?**

```python
# Subset all rows where the start_station is 8 Ave & W 31 St.
station_8AveW31St = df_clean.query('start_station_name == "8 Ave & W 31 St"')
station_8AveW31St.shape
```

```python
# count the number of rides based on the end_station_name.
num_end_stations = station_8AveW31St.groupby('end_station_name')['starttime'].nunique().reset_index(name='num_trips').sort_values(by='num_trips', ascending=False)
num_end_stations.head(10)
```

```python
# Observe the distributions of the trips.
sns.histplot(num_end_stations, x='num_trips', bins=20, kde=True)
```

## **Which months of the year has the highest demand for bikes?**

```python
# calculate the number of rides per month in 2014 and plot it.
trips_month = df_clean.groupby('months')['starttime'].nunique().reset_index(name='monthly_trips')
trips_month
```

```python
# convert the months feature to names.
trips_month['months'] = pd.to_datetime(trips_month['months'], format='%m').dt.month_name()
trips_month
```

```python
# plot the above data using line graph.
plt.figure(figsize=(5,4))
sns.lineplot(trips_month, y='monthly_trips', x='months', color='red')
plt.xticks(rotation=90)
plt.ylim(0,)
plt.show()
```

### As expected the months between May and October the demand for bike is high. Likely due to good weather.

## **Are there pick hours for bike demand?**

```python
# Count the number of bikes used by each hour.
ride_hours = df_clean.groupby('hour')['starttime'].nunique().reset_index(name='num_trips')
ride_hours
```

```python
plt.figure(figsize=(6,4))
sns.lineplot(ride_hours, x='hour', y='num_trips', color='red')
plt.xticks(range(0, 25), rotation=90)
plt.show()
```

### The bike usage is generally high during the day. Particularly in the morning between 8 and 9 AM and between 5 and 7PM the demands are quite high. This likely be due to the work hours.
